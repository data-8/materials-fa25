## Rolling dice

#### Task

- Simulate and store the results of rolling a fair, six-sided dice 100 times in a one-column table in two ways:
    - Simulating from an array that contains the sides of the die
    - Simulating from a table that contains the sides of the die

```python
die_numbers = make_array()
for i in np.arange(100):
    new_die_roll = np.random.choice(make_array(1,2,3,4,5,6))
    die_numbers = np.append(new_die_roll, die_numbers)
```

```python
die_numbers = make_array()
die_table = Table().with_columns('Sides', np.arange(1,7))

for i in np.arange(100):
    new_die_roll = die_table.sample(1).column('Sides').item(0)
    die_numbers = np.append(new_die_roll, die_numbers)
```

```python
die_numbers
```

```python
die_table = Table().with_columns('Result', (die_numbers))
```

```python
die_table.hist('Result', bins = make_array(0,3,5,6,7), unit = 'Spot')
```

## Restaurant growth report

#### Task

- Read in and store the data in the visualization from the file `restaurant_growth.csv.`

```python
rg = Table.read_table("restaurant-growth.csv")
```

#### Task

- Add the 'Transactions YOY' column to the table that you see in the visualization using two different methods!
- See if you can use the text to help you (or, just stare at the columns and you'll figure out what to do)!

```python
rg = rg.with_columns('Transactions YOY', 
                (rg.column('2024')-rg.column('2023'))/rg.column('2023')*100)

rg
```

```python
def percent_growth(initial, changed):
    return (changed-initial)/initial * 100
```

```python
rg = rg.with_columns('Transactions YOY', rg.apply(percent_growth, '2023', '2024'))
rg
```

#### Task

- Visualize the distribution of the year-over-year transactions using the plot type in the visualization, but make it easier to see which restaurants saw the most growth.

```python
rg.sort('Transactions YOY', descending=True).select('City','Transactions YOY').barh('City')
```
