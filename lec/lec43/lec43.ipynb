## Rolling dice

#### Task [5 min]

- Simulate and store the results of rolling a fair, six-sided die 100 times in a one-column table in two ways:
    - Simulating from an array that contains the sides of the die
    - Simulating from a table that contains the sides of the die

```python
die_results = make_array()
```

```python
die = Table().with_columns('Sides',make_array(1,2,3,4,5,6))
die
```

```python
die.sample(1).column('Sides').item(0)
```

```python
for i in np.arange(100):
    new_roll = die.sample(1).column('Sides').item(0)
    die_results = np.append(die_results, new_roll)
```

```python
die_results
```

```python
die_table = Table().with_columns('Results', die_results)
die_table
```
